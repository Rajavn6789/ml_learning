### HOF 
Higher order functions

```python
def make_adder(n): 
    def add(x):
        return x + n      
    return add 
    
add_five = make_adder(5) 

result = add_five(3)   # returns 8
print(result)
```

### List 
List operations

```python
numbers = [1, 2, 3, 4, 5]

# Transformations
squares = [x ** 2 for x in numbers]
cubes = [x ** 3 for x in numbers]
double = [x * 2 for x in numbers]
half = [x / 2 for x in numbers]

# Filtering
evens = [x for x in numbers if x % 2 == 0]
odds = [x for x in numbers if x % 2 != 0]
greater_than_3 = [x for x in numbers if x > 3]
less_than_4 = [x for x in numbers if x < 4]

# Conditional expressions
labels = ["Even" if x % 2 == 0 else "Odd" for x in numbers]
pass_fail = ["Pass" if x >= 3 else "Fail" for x in numbers]

# Built-in functions
total = sum(numbers)
largest = max(numbers)
smallest = min(numbers)
count = len(numbers)
average = sum(numbers) / len(numbers)

# Reverse
reversed_list = numbers[::-1]

# Sorting
descending = sorted(numbers, reverse=True)

# Copy
copy_list = numbers.copy()

# List methods
numbers.append(6)
numbers.insert(0, 0)
numbers.remove(3)
last = numbers.pop()

# Membership
has_four = 4 in numbers
has_ten = 10 in numbers

# Enumerate
indexed = [(i, x) for i, x in enumerate(numbers)]

# Zip
letters = ["A", "B", "C", "D"]
paired = list(zip(letters, numbers))

# Flatten a nested list
nested = [[1, 2], [3, 4], [5, 6]]
flat = [item for sublist in nested for item in sublist]

# Print results
print("numbers:", numbers)
print("squares:", squares)
print("cubes:", cubes)
print("double:", double)
print("half:", half)

print("evens:", evens)
print("odds:", odds)
print("greater_than_3:", greater_than_3)
print("less_than_4:", less_than_4)

print("labels:", labels)
print("pass_fail:", pass_fail)

print("sum:", total)
print("max:", largest)
print("min:", smallest)
print("count:", count)
print("average:", average)

print("reversed:", reversed_list)
print("descending:", descending)
print("copy:", copy_list)

print("has_four:", has_four)
print("has_ten:", has_ten)

print("indexed:", indexed)
print("paired:", paired)
print("flat:", flat)
```

### Generators 
A generator is a function that produces values lazily, one at a time on demand, instead of all at once in memory.

```python
def count_up(n):
    i = 0
    while i < n:
        yield i
        i += 1 
        
for x in count_up(3):      
    print(x)   # prints 0, 1, 2
```

### Dunder Methods   
Python allows classes to define how they behave with built-in operators. Methods named with double underscores (__add__, __mul__, __repr__) are called dunder methods (short for "double underscore"). They are the machinery behind operator overloading.

```python
class Box:      
    def __init__(self, size):          
        self.size = size      
    def __add__(self, other):          
        return Box(self.size + other.size)      
    def __repr__(self):          
        return f"Box(size={self.size}) mm"

a = Box(50)
b = Box(50)
c = a + b
print(c)
```
