## Attention

```python
import numpy as np
from gensim.models import Word2Vec
```

```python
np.random.seed(42)
d_model = 4
d_k = 3

def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)
```

```python
# Embeddings
sentences = [
    ["cat", "eats", "fish"]
]

model = Word2Vec(
    sentences,
    vector_size=d_model,
    window=2,
    min_count=1,
    workers=1
)
```

```python
X = np.array([model.wv[word] for word in ["cat", "eats", "fish"]])

print("Input matrix X:")  
print(X)  
print(f"Shape: {X.shape}")  
```

```python
Wq = np.random.randn(d_model, d_k) * 0.1  
Wk = np.random.randn(d_model, d_k) * 0.1  
Wv = np.random.randn(d_model, d_k) * 0.1

print(f"Wq Shape: {Wq.shape}")  
print(f"Wk Shape: {Wk.shape}")  
print(f"Wv Shape: {Wv.shape}")  
```

```python
# Create Query matrix  
Q = np.dot(X, Wq)  
print("Query matrix Q:")  
print(Q)  
print(f"Shape: {Q.shape}")  
```

```python
# Create Key matrix  
K = np.dot(X, Wk)  
print("Keys matrix K:")  
print(K)  
print(f"Shape: {K.shape}")  
```

```python
# Create Value matrix  
V = np.dot(X, Wv)  
print("Value matrix K:")  
print(V)  
print(f"Shape: {V.shape}")  
```
