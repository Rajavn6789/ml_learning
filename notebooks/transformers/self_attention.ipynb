```python
import torch
import torch.nn as nn
import math
```

```python
#1. vocabulary
vocab = {
    "<pad>": 0,
    "dog": 1,
    "chasing": 2,
    "cat": 3     
}

sentence1 = ["dog","chasing", "cat"]
sentence2 = ["cat",  "chasing", "dog"]

def tokenize(sentence):
    return torch.tensor([vocab[word] for word in sentence])

    
tokens1 = tokenize(sentence1)
tokens2 = tokenize(sentence2)

print(tokens1)
print(tokens2)
```

```python
# 2 Word Embedding Layer
embedding_dim = 4

embedding = nn.Embedding(
    num_embeddings = len(vocab),
    embedding_dim = embedding_dim
)

x1 = embedding(tokens1)
x2 = embedding(tokens2)


print("\nWord embeddings1")
print(x1)

print("\nWord embeddings2")
print(x2)
```

```python
Wq = nn.Linear(embedding_dim, embedding_dim, bias=False)
Wk = nn.Linear(embedding_dim, embedding_dim, bias=False)
Wv = nn.Linear(embedding_dim, embedding_dim, bias=False)


def self_attention(x):

    Q = Wq(x)
    K = Wk(x)
    V = Wv(x)

    # Attention scores
    scores = Q @ K.T

    # Normalize
    weights = torch.softmax(scores, dim=-1)

    # Weighted sum
    output = weights @ V

    return output, weights
```

```python
out1, attn1 = self_attention(x1)
out2, attn2 = self_attention(x2)

print(attn1)

print(attn2)
```
