```python
import numpy as np
```

```python
groupA = np.array([6, 1]) 
groupB = np.array([1, 10]) 
groupC = np.array([7, 7])
```

```python
probA0 = groupA[0] / groupA.sum()
probA1 = groupA[1] / groupA.sum()
surpriseA0 = -np.log2(probA0)
surpriseA1 = -np.log2(probA1)
entropyA = probA0 * surpriseA0 + probA1 * surpriseA1


probB0 = groupB[0] / groupB.sum()
probB1 = groupB[1] / groupB.sum()
surpriseB0 = -np.log2(probB0)
surpriseB1 = -np.log2(probB1)
entropyB = probB0 * surpriseB0 + probB1 * surpriseB1


probC0 = groupC[0] / groupC.sum()
probC1 = groupC[1] / groupC.sum()
surpriseC0 = -np.log2(probC0)
surpriseC1 = -np.log2(probC1)
entropyC = probC0 * surpriseC0 + probC1 * surpriseC1
```

```python
import pandas as pd

rows = [
    ("A", "orange", int(groupA[0]), probA0, surpriseA0),
    ("A", "blue",   int(groupA[1]), probA1, surpriseA1),
    ("B", "orange", int(groupB[0]), probB0, surpriseB0),
    ("B", "blue",   int(groupB[1]), probB1, surpriseB1),
    ("C", "orange", int(groupC[0]), probC0, surpriseC0),
    ("C", "blue",   int(groupC[1]), probC1, surpriseC1),
]

table = pd.DataFrame(rows, columns=["Group", "Class", "count", "probability", "surprise"])
table["probability"] = table["probability"].round(3)
table["surprise"] = table["surprise"].round(3)

print(table.to_string(index=False))
```

```python
entropy_table = pd.DataFrame(
    [("A", entropyA), ("B", entropyB), ("C", entropyC)],
    columns=["Group", "entropy"],
)
entropy_table["entropy"] = entropy_table["entropy"].round(3)

print(entropy_table.to_string(index=False))
```
