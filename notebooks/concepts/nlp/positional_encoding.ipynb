```python
import torch
import torch.nn as nn
import math
```

```python
#1. vocabulary
vocab = {
    "<pad>": 0,
    "dog": 1,
    "is": 2,
    "chasing": 3,
    "cat": 4     
}

sentence1 = ["dog", "is", "chasing", "cat"]
sentence2 = ["cat", "is", "chasing", "dog"]

def tokenize(sentence):
    return torch.tensor([vocab[word] for word in sentence])

    
tokens1 = tokenize(sentence1)
tokens2 = tokenize(sentence2)

print(tokens1)
print(tokens2)
```

```python
# 2 Word Embedding Layer

embedding_dim = 8

embedding = nn.Embedding(
    num_embeddings = len(vocab),
    embedding_dim = embedding_dim
)

word_embeddings1 = embedding(tokens1)
word_embeddings2 = embedding(tokens2)

print("\nWord embeddings1 shape")
print(word_embeddings1.shape)

print("\nWord embeddings2 shape")
print(word_embeddings2.shape)


print("\nWord embeddings1")
print(word_embeddings1)

print("\nWord embeddings2")
print(word_embeddings2)
```

```python
pe = torch.zeros(10, 8)
position = torch.arange(0, 10).unsqueeze(1)
```

```python
even_dimesion_indices = torch.arange(0, 8, 2)
decay_factor = -math.log(10000)/8


div_term = even_dimesion_indices * decay_factor
exp_div_term = torch.exp(div_term)
print(div_term, exp_div_term)
```

```python
pe[:, 0::2] = torch.sin(position * exp_div_term)
pe[:, 1::2] = torch.cos(position * exp_div_term)
```
