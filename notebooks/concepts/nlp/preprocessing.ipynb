```python
import re
import pandas as pd
from nltk import pos_tag
from nltk.corpus import stopwords, wordnet
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
countvectorizer = CountVectorizer()
tfidvectorizer = TfidfVectorizer()
```

## Basic Concepts

### 1. Text Cleaning

```python
review = """I've been using this wireless mouse for about 3 weeks now, and honestly,
it's exceeded my expectations! The battery life is incredible - I charged it once
and it's still going strong. It's super comfortable, even during those 10-hour
work-from-home marathons, and the connection hasn't dropped a single time.

That said, it wasn't all perfect. The packaging arrived slightly damaged (the box
was crushed on one corner), and the scroll wheel felt a bit stiff for the first
couple of days. Customer support was responsive though - they replied within 24 hrs
and offered a 20% refund. Would I buy it again? Absolutely. 5/5 stars!!!"""


reviews = [
    "This product is amazing",
    "Amazing quality and great value",
    "Poor quality and bad experience"
]


cleaned = re.sub(r"[^a-zA-Z0-9/\s']", '', review)
print(cleaned)
```

### 2. Tokenisation

```python
tokens = word_tokenize(review)
print(tokens)
```

### 3. Normalisation

```python
tokens = [word.lower() for word in tokens  if word.isalpha()]
print(tokens)
```

### 4. Stopword Removal

```python
tokens = [
    word for word in tokens
    if word not in stop_words
]
print(tokens)
```

### 5. Lemmatization

```python
tokens = [lemmatizer.lemmatize(word, pos='v')
              for word in tokens]
print(tokens)
```

## Advanced Concepts

### 1. Bag of Words

```python
X = countvectorizer.fit_transform(reviews)

bow_df = pd.DataFrame(
    X.toarray(),
    columns=countvectorizer.get_feature_names_out()
)

bow_df
```

### 2. TFID vectorizer

```python

X = tfidvectorizer.fit_transform(reviews)

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=tfidvectorizer.get_feature_names_out()
)

tfidf_df
```

### 3. N grams

```python
from sklearn.feature_extraction.text import CountVectorizer

unigramvectorizer = CountVectorizer(ngram_range=(1,1))

unigramvectorizer.fit_transform(reviews)

print(unigramvectorizer.get_feature_names_out())


bigramvectorizer = CountVectorizer(ngram_range=(2,2))

bigramvectorizer.fit_transform(reviews)

print(bigramvectorizer.get_feature_names_out())
```

### 4. Regex-Based Feature Extraction

```python
features = {
    # Numbers
    "num_count": len(re.findall(r'\d+', review)),

    # Percentages
    "percentage_count": len(re.findall(r'\d+%', review)),

    # Ratings like 5/5
    "rating_count": len(re.findall(r'\d+/\d+', review)),

    # Time expressions
    "weeks_count": len(re.findall(r'\b\d+\s*weeks?\b', review, re.I)),
    "hours_count": len(re.findall(r'\b\d+\s*hours?\b', review, re.I)),
    "hrs_count": len(re.findall(r'\b\d+\s*hrs?\b', review, re.I)),

    # Exclamation marks
    "exclamation_count": len(re.findall(r'!', review)),

    # Question marks
    "question_count": len(re.findall(r'\?', review)),

    # ALL CAPS words
    "all_caps_count": len(re.findall(r'\b[A-Z]{2,}\b', review)),

    # Hyphenated words
    "hyphenated_words": len(re.findall(r'\b\w+(?:-\w+)+\b', review)),

    # Contractions
    "contractions": len(re.findall(r"\b\w+'\w+\b", review)),

    # Positive sentiment words
    "positive_words": len(re.findall(
        r'\b(amazing|incredible|comfortable|responsive|absolutely|great|excellent)\b',
        review,
        re.I
    )),

    # Negative sentiment words
    "negative_words": len(re.findall(
        r'\b(damaged|crushed|stiff|bad|poor|terrible)\b',
        review,
        re.I
    )),

    # Refund mentions
    "refund_mentions": len(re.findall(r'\brefund\b', review, re.I)),

    # Customer support mentions
    "support_mentions": len(re.findall(r'\bsupport\b', review, re.I)),

    # Word count
    "word_count": len(re.findall(r'\b\w+\b', review)),

    # Sentence count
    "sentence_count": len(re.findall(r'[.!?]+', review)),
}


print(features)
```
