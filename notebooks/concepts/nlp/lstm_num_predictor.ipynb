```python
import torch
import torch.nn as nn
```

```python
# Training data
X = torch.tensor([
    [1, 2, 3],
    [2, 3, 4],
    [3, 4, 5],
    [4, 5, 6],
    [5, 6, 7],
    [10, 901, 902]
], dtype=torch.float32)

y = torch.tensor([
    [4],
    [5],
    [6],
    [7],
    [8],
    [903]
], dtype=torch.float32)
```

```python
# LSTM expects:
# (batch_size, sequence_length, features)
X = X.unsqueeze(-1)
```

```python
class NumberLSTM(nn.Module):
    def __init__(self, hidden_size=10):
        super().__init__()

        #lstm
        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=hidden_size,
            batch_first=True
        )

        # fc
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        output, (hidden, cell) = self.lstm(x)

        # Use final hidden state
        return self.fc(hidden[-1])
```

```python
model = NumberLSTM()
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
for epoch in range(500):
    pred = model(X)
    loss = criterion(pred, y)
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    if epoch % 50 == 0:
        print(f"Epoch {epoch}, Loss: {loss.item():.4f}")
```

```python
# Predict next number after 6,7,8
test = torch.tensor([[10, 11, 12]], dtype=torch.float32)
test = test.unsqueeze(-1)

prediction = model(test)

print("\nPrediction:", prediction.item())
print("Expected: 9")
```
