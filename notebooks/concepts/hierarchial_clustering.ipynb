```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
# Customer data
data = pd.DataFrame({
    "Age": [25, 27, 45, 46, 35, 36],
    "Income": [30, 32, 90, 95, 60, 62],
    "Spending": [80, 78, 20, 18, 50, 52]
})
```

```python
# ------------------------
# Step 1: Build Dendrogram
# ------------------------
linked = linkage(data, method='ward')

names = ['A','B','C','D','E','F']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,4))

# Left: dendrogram
dendrogram(
    linked,
    labels=names,
    ax=ax1
)
ax1.set_title("Customer Dendrogram")
ax1.set_xlabel("Customers")
ax1.set_ylabel("Distance")

# Right: scatter of the raw customers
ax2.scatter(data["Income"], data["Spending"])
for i, name in enumerate(names):
    ax2.annotate(name, (data["Income"][i], data["Spending"][i]))
ax2.set_title("Customers: Income vs Spending")
ax2.set_xlabel("Income")
ax2.set_ylabel("Spending")

plt.tight_layout()
plt.show()
```

```python
# ------------------------
# Step 2: Create Clusters
# ------------------------
hc = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

clusters = hc.fit_predict(data)

data["Cluster"] = clusters

print(data)
```
