## Normal Distribution of Members

A histogram shows how many members fall into each height range. Overlaying a **Normal (Gaussian) curve** built from the data's own mean and standard deviation lets us see how closely the heights follow a bell shape.

- The **blue bars** are the actual (empirical) distribution of heights.
- The **red curve** is the theoretical Normal distribution `N(μ, σ²)`.
- The dashed lines mark the **mean** and the **±1σ / ±2σ** bands (≈68% of data within ±1σ, ≈95% within ±2σ).

```python
import matplotlib.pyplot as plt
from scipy.stats import norm

data = np.array(members)
mu = data.mean()
sigma = data.std()
variance = data.var()

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Left: Normal distribution (histogram + fitted Normal curve)
axes[0].hist(data, bins=15, density=True, color="steelblue",
             edgecolor="white", alpha=0.7, label="Members (empirical)")
x = np.linspace(data.min() - 5, data.max() + 5, 300)
axes[0].plot(x, norm.pdf(x, mu, sigma), color="crimson", lw=2.5,
             label=f"Normal N(μ={mu:.1f}, σ={sigma:.1f})")
axes[0].axvline(mu, color="black", ls="--", lw=1.5, label="Mean")
for k in (1, 2):
    axes[0].axvline(mu + k * sigma, color="gray", ls=":", lw=1)
    axes[0].axvline(mu - k * sigma, color="gray", ls=":", lw=1)
axes[0].set_title("Normal Distribution of Member Heights")
axes[0].set_xlabel("Height (cm)")
axes[0].set_ylabel("Probability density")
axes[0].legend()

# Right: Deviations from the mean (variance = mean of squared deviations)
order = np.argsort(data)
sorted_data = data[order]
axes[1].vlines(range(len(data)), mu, sorted_data, color="gray", lw=0.8)
axes[1].scatter(range(len(data)), sorted_data, s=12, color="crimson")
axes[1].axhline(mu, color="black", lw=1.5, label=f"Mean = {mu:.1f}")
axes[1].set_title(f"Deviations from mean  (σ²={variance:.1f}, σ={sigma:.1f})")
axes[1].set_xlabel("Members (sorted by height)")
axes[1].set_ylabel("Height (cm)")
axes[1].legend()

plt.tight_layout()
plt.show()
```
