```python
import torch
```

```python
B = 3
T = 4
C = 8
H = 2
D = C // H
```

```python
torch.manual_seed(32)
x = torch.randint(high=99, size=(B, T, C))
print(x)
print("Original:", x.shape)
```

```python
# Split heads
x = x.view(B, T, H, D)
print("Split:", x.shape) # (3, 4, 2, 4)
print(x)
```

```python
# Move heads before tokens
x = x.transpose(1, 2) 
print("Attention format:", x.shape) # (3, 2, 4, 4)
print(x)
```

```python
# ---- MERGE HEADS ----

# B H T D -> B T H D
x = x.transpose(1, 2)
print("Before merge:", x.shape)
print(x)
```

```python
# B T H D -> B T C
x = x.reshape(B, T, C)

print("Merged:", x.shape)
print(x)
```
