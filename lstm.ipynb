```python
import torch
import math
import numpy as np
```

```python
# Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Tan H
def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
```

```python
# Data
C_t_minus_1 = 1.5  
H_t_minus_1 = 0.9 
X_t = 1

inputs = np.array([H_t_minus_1, X_t])
```

```python
# Weights and Biases
W_f = np.array([2.70, 1.63])
W_i = np.array([2.0, 1.65])
W_c = np.array([1.41, 0.94])
W_o = np.array([4.38, -0.19])

b_f = 1.62
b_i = 0.62
b_c = -0.32
b_o = 0.59
```

```python
#Forget Gate
F_t = sigmoid(np.dot(W_f, inputs) + b_f)
print("F_t", F_t)
```

```python
# Update long term memory
C_t_minus_1 = C_t_minus_1 * F_t 
print("C_t_minus_1", C_t_minus_1)
```

```python
# Input Gate
I_t = sigmoid(np.dot(W_i, inputs) + b_i)
print("I_t", I_t)

# Candidate  memory
C_t_bar = tanh(np.dot(W_c, inputs) + b_c)
print("C_t_bar", C_t_bar)
```

```python
# Update long term memory
C_t =(C_t_bar * I_t) +  C_t_minus_1
print("C_t", C_t)
```

```python
O_t = sigmoid(np.dot(W_o, inputs) + b_o)
print("O_t", O_t)
```

```python
# Update short term memory
H_t = O_t *  tanh(C_t)
print("H_t", H_t)
```
