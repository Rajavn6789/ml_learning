```python
import torch
import torch.nn as nn
```

```python
text = "hello world"
```

```python
# -------------------------
# 1. Character vocabulary
# -------------------------
chars = sorted(set(text))

stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for ch, i in stoi.items()}

vocab_size = len(chars)

print(chars)
print("vocab size:", vocab_size)
```

```python

# -------------------------
# 2. Convert text to IDs
# -------------------------
data = torch.tensor([
    stoi[ch] for ch in text
])

print("IDs:")
print(data)

print("\nShape:")
print(data.shape)
```

```python

# -------------------------
# 3. Embedding
# -------------------------
EMBED_DIM = 8

embedding = nn.Embedding(
    vocab_size,
    EMBED_DIM
)

x = data[]

print("\nInput IDs:")
print(x)

vectors = embedding(x)

print("\nEmbedding:")
print(vectors)

print("\nShape:")
print(vectors.shape)
```
